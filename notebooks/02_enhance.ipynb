# Marine video enhancement

Cleans an aerial marine clip: glint/foam mask + Telea inpainting, edge-preserving smoothing, CLAHE on the LAB lightness channel, then encodes a clean H.264 MP4 for the downstream SAM3/DINOv3 notebooks.

Note: the `fps` argument is a fallback only — a valid source FPS overrides it.

```
from fish_segmentation.enhance import process_marine_video_ffmpeg
from fish_segmentation.paths import find_repo_root, load_env

load_env()
ROOT = find_repo_root()

OUTPUTS = ROOT / "notebooks" / "outputs"
OUTPUTS.mkdir(parents=True, exist_ok=True)
```

```
input_video = ROOT / "data" / "test_media" / "dolphin_00.mp4"
output_video = OUTPUTS / "preprocessed.mp4"

process_marine_video_ffmpeg(str(input_video), str(output_video), fps=15.0)
```
